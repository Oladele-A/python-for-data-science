### Exercise on pandas series

```python
# Given a list representing a few planets
planets = ['Earth', 'Saturn', 'Venus', 'Mars', 'Jupiter']

# Given another list representing the distance of each of these planets from the Sun
# The distance from the Sun is in units of 10^6 km
distance_from_sun = [149.6, 1433.5, 108.2, 227.9, 778.6]

# TO DO: Create a Pandas Series "dist_planets" using the lists above, representing the distance of the planet from the Sun.

# TO DO: Calculate the time (minutes) it takes light from the Sun to reach each planet. 
# You can do this by dividing each planet's distance from the Sun by the speed of light.
# Use the speed of light, c = 18, since light travels 18 x 10^6 km/minute.

# TO DO: Use Boolean indexing to select only those planets for which sunlight takes less
# than 40 minutes to reach them.
```

```python
dist_planets = pd.Series(data = distance_from_sun, index = planets)
dist_planets
```

```python
speed_of_light = 18
light_time = dist_planets / speed_of_light
light_time
```

```python
close_planets = light_time[light_time < 40]
close_planets
```

### Handling NaN Values -  Method 1: Remove NaN

```python
# Eliminate NaN values using dropna method

store_inventories.dropna(axis = 0) # Remove all rows with NaN. Out of place
```

```python
store_inventories.dropna(axis = 1) # Remove all columns with NaN. Out of place
```

```python
store_inventories
```

```python
copy_store = store_inventories.copy()
copy_store
```

```python
store_inventories
```

```python
store_inventories.fillna(0) # Replace NaN with 0. Out of place. Set inplace parameter to true to fill in place
```

```python
store_inventories
```

```python
store_inventories.fillna(method='ffill', axis=0) # Replace NaN with the value from the preceding row. Similar to filldown in Excel
```

```python
store_inventories.ffill(axis=0) # Replace NaN with the value from the preceding row. Similar to filldown in Excel
```

```python
store_inventories.fillna(method='ffill', axis =1) # Replace NaN with the value from the preceding column.
```

```python
store_inventories.ffill(axis=1) # Replace NaN with the value from the preceding column.
```

```python
store_inventories.bfill(axis=1) # Replace NaN with the value from the next column.
```

```python
# TO DO: Create a Pandas DataFrame using the dictionary created above

df = pd.DataFrame(dict)
df
```

```python
# TO DO:
# If you created the dictionary correctly you should have a Pandas DataFrame
# that has column labels: 
# 'Author', 'Book Title', 'User 1', 'User 2', 'User 3', 'User 4' 
# and row indices 0 through 4.

# Now replace all the NaN values in your DataFrame with the average rating in
# each column. Replace the NaN values in place. 
# HINT: Use the `pandas.DataFrame.fillna(value, inplace = True)` function for substituting the NaN values. 
# Write your code below:

df.fillna(df.mean(axis=0, numeric_only=True), inplace=True)
df
```
