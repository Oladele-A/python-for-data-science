### Drop Missing Data

```python
df
```

```python
df.dropna() #drop all rows that contain null values
```

```python
df.dropna(thresh=1) #Keep rows that contain at least (the value of thresh) non-null values
```

```python
df.dropna(axis=1) #drop all columns that contain an instance of a missing value
```

```python
df.dropna(subset=['last_name']) #drop all rows that contain a missing value under the given column(s)
```

```python
year_mod.xs(key=4, level='cylinders') # All cylinder components under the given key
```
