```python
for word in book_title:
    if word not in word_counter:
        word_counter[word] = 1
    else:
        word_counter[word] += 1

print(word_counter)
```

```python
cargo_weight = 0
cargo_items = []

for item, weight in manifest:
    if cargo_weight >= 100:
        break # to break out of the loop when a condition is met
    elif cargo_weight + weight > 100:
        continue # to skip an iteration that does/does not meet a certain condition
    else:
        cargo_items.append(item)
        cargo_weight += weight

print(cargo_weight, cargo_items)
```

## List Comprehension

```python
# Recall this example
cities = ['lagos', 'accra', 'nantes', 'zurich', 'bern', 'paris']
capitalized_cities = []
for city in cities:
    capitalized_cities.append(city.title())

print(capitalized_cities)
```

```python
# Solution with list comprehension
cities = ['lagos', 'accra', 'nantes', 'zurich', 'bern', 'paris']
capitalized_cities = [city.title() for city in cities]
print(capitalized_cities)
```

```python
squares = []
for x in range(9):
    squares.append(x**2)
print(squares)

# With List Comprehension
squares = [x**2 for x in range(9)]
print(squares)
```

```python
# List Comprehension with if statement
# When using only the if statement, the if statement comes after the iterable. 
# This if statement acts as a filter for the loop
# Squares of even number. 

even_squares = [x**2 for x in range(9) if x % 2 == 0]
print(even_squares)
```

```python
# List Comprehension with if else conditional statement
# When using the if..else conditional, The if else statement must come before the iterable in a list comprehension
# In this case, the if else statement acts a conditional expression that evaluates every single iteration. 

another = [x**2 if x % 2 == 0 else x + 3 for x in range(9)]
print(another)
```
