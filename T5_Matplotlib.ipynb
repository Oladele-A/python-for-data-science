## Visualisation Using Plot

```python
plt.plot(x,y)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.xlim(0,10)
plt.ylim(0,20)
plt.title('String Title');
# plt.savefig(r"C:\Users\Olade\Desktop\my_first_plot.png");
```

## Visualisation Using Figure

```python
# Create a background figure and then define an axis on that figure
fig = plt.figure()
axes = fig.add_axes([0,0,1,1]) #The 1st two arguments refer to the location on the figure. The last two form the % size of the axis wrt the fig
```

```python
a = np.linspace(0,10, 11)
b = a**4
x = np.arange(0,10)
y = 2*x
```

```python
a 
```

```python
b
```

```python
x
```

```python
y
```

```python
fig = plt.figure()

# Large Axes
axes1 = fig.add_axes([0,0,1,1])
axes1.plot(a,b)
axes1.set_xlim(0,12)
axes1.set_ylim(0,12000)
axes1.set_xlabel('A')
axes1.set_ylabel('B')
axes1.set_title('Power of 4')

# Small Axes
axes2 = fig.add_axes([0.2,0.5, 0.25,0.25])
axes2.plot(a,b)
axes2.set_xlim(0,2)
axes2.set_ylim(0,48)
axes2.set_xlabel('X')
axes2.set_ylabel('Y')
axes2.set_title('Zoomed In');
```

### Visualisation Using Subplots

```python
print(a)
print(b)
print(x)
print(y)
```

```python
fig, ax = plt.subplots(2,2)

ax[0,0].plot(x,y)
ax[0,0].set_xlabel('X')
ax[0,0].set_ylabel('Y')
ax[0,0].set_title('Multiples of 2')

ax[1,0].plot(a,b)
ax[1,0].set_xlabel('A')
ax[1,0].set_ylabel('B')
ax[1,0].set_title('Power of 4')

# Set figure-level Parameter
fig.suptitle('Figure Level Title',fontsize=10)

# Automatically adjust the space between axes
# plt.tight_layout() 

# Manually Adjust the space between axex
fig.subplots_adjust(hspace=0.5, wspace=0.5)

# Save the figure
# fig.savefig(r"C:\Users\Olade\Desktop\new_multiple_fig.png",bbox_inches='tight');
```

## Adding legend to a Plot

```python
x = np.linspace(0,10,11)
x
```

```python
fig = plt.figure(figsize=(3,2))
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x, label='X vs X')
ax.plot(x,x**2, label='X vs X^2')
ax.legend(loc=(1.1, 0.5));
```

## Other Plot Styling

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x,color = '#8c069e', lw=1, ls='--', marker='D', ms=5) # Hex Color Picker
ax.plot(x,x+2, color ='orange', linewidth=2, linestyle=':', marker='o', markersize=10) # Marker string code


# lw = linewidth, ls = linestyle, ms = markersize
# Possible linestyle options '--', '-', '-.', ':', 'dashed', 'dashdot', 'solid', 'None'
```
