### Q1 - Show the relationship between the days employed versus the age of the person (DAYS_BIRTH) for people who were not unemployed.

```python
employed_df = credit_df[credit_df['DAYS_EMPLOYED'] < 0].copy()
```

```python
employed_df.head()
```

```python
employed_df['DAYS_EMPLOYED'] = employed_df['DAYS_EMPLOYED'] * -1

employed_df['DAYS_BIRTH'] = employed_df['DAYS_BIRTH'] * -1
```

```python
plt.figure(figsize=(9,4), dpi=150)
sns.scatterplot(data=employed_df, x='DAYS_BIRTH', y='DAYS_EMPLOYED', alpha=0.02, linewidth=0);
```

### Q2 - Show the Distribution of ages in years

```python
credit_df['YEARS'] = -1 * credit_df['DAYS_BIRTH']/365
```

```python
sns.histplot(data=credit_df, x='YEARS', bins=30, color='red', edgecolor='black', alpha=0.4);
```

### Q3 - Show the distribution of total income per family status for the bottom half earners

```python
n_half = int(len(credit_df)/2)
```

```python
bottom_half = credit_df.nsmallest(n=n_half, columns='AMT_INCOME_TOTAL')
```

```python
plt.figure(figsize=(12,5), dpi=150)
sns.boxplot(data=bottom_half, x='NAME_FAMILY_STATUS', y='AMT_INCOME_TOTAL', hue= 'FLAG_OWN_REALTY')
plt.legend(loc=(1.05,0.5), title='FLAG_OWN_REALTY')
plt.ylim(10000,180000)
plt.title('Income Totals per Family Status for Bottom Half Earners');
```
