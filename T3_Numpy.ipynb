### Deleting Elements of an Array
The delete function takes an **array**, the **indices** to be deleted and the **axis** to delete from as arguments.
0 = Row axis, 1 = column axis. The axis can be omitted for a rank 1 array

```python
#  Rank 1 Array
x = np.array([1,2,3,4,5])
print(x)

x = np.delete(x,[0,4])
print(x)
```

```python
# Rank 2 Array
y = np.arange(1,10).reshape(3,3)
print(y, '\n')

n = np.delete(y,[0], axis = 0)
print(n, '\n')

v = np.delete(y,[0,2], axis = 1)
print(v, '\n')
```

### Appending Elements to an Array
The append function takes an **array**, a list of **elements** to append, and the **axis** to append on as parameters

```python
#  Rank 1 Array
x = np.array([1,2,3,4,5])
print(x)

x = np.append(x,[6]) # The axis argument can be omitted
print(x)
```

```python
x = np.append(x,[7,8], axis = 0)
print(x)
```

```python
# Rank 2 Array
y = np.arange(1,10).reshape(3,3)
print(y)
```

# Stacking Numpy Arrays


```python
a = np.array([[1,2,3]])
a
# a.ndim
```

```python
b = np.append(a,[[4,5,6]], axis = 0)
b
# b.ndim
```

```python
# Consider the example above
# Array a is already a rank 2 array, so appending b along axis 0 looks like it is stacked below a. 
# This will not be the case if array a is a rank 1 array. That is where stacking comes in. 
a= np.array([1,2,3])
a
b =np.append(a,[4,5,6],axis =0)
b
```

### Stacking helps to change the dimension/rank/structure of an array by introducing a new axis.
### For example, you can start from 1-D arrays and stack them to produce a 2-D (rank-2) array.
### Note that the shape of the arrays must be the same before stacking

```python
x = np.array([1,2])
print(x)
print("Shape: " , x.shape)
```

```python
y = np.array([[3,4], [5,6]])
print(y)
print('Shape: ', y.shape)
```

```python
z = np.vstack((x,y))
print(z)
print('Shape: ', z.shape)
print('Size: ', z.size)
```

```python
w = np.hstack((y,x.reshape(2,1)))
print(w)
```

# Slicing ndarrays
For multidimensional arrays, you will have to specify a slice for each dimension of the array

### Methods
**ndarray[start : end]**
**ndarray[: end]**
**ndarray[start : ]**

```python
x = np.arange(1,21).reshape(4,5)
print(x)
```

```python
z = x[1:4, 2:5]
print(z, '\n')

z = x[1:, 2:]
print(z)
```

```python
y = x[:3,2:]
print(y)
```

```python
w = x[:,2:3] # Give me the column while preserveing the dimension
print(w)
```

```python
w = x[:,2] # Give me the values & don't worry about dimensions
print(w)
```

### Note that slicing numpy arrays does not copy the slice into a new variable. 
Slicing only creates a view of the intial array. Thus, changing a value in the new variable will affect the initial variable

```python
x = np.arange(1,21).reshape(4,5)
x
```

```python
z = x[1:,2:]
z
```

```python
z[2,2] = 555
z
```

```python
# Z is just a view of x. Changing any value in z will change the same value in x
print(x) 
```

### Copying an array
The copy command creates a new numpy array that is completely independent of the original

```python
x = np.arange(1,21).reshape(4,5)
x
```

```python
z = np.copy(x[1:,2:])
print(z, '\n')

y = x[:3,2:].copy() # The 2 methods of copying
print(y)
```

```python
z[2,2] = 555
print(z, '\n') # Here, modifying z does not affect the value of x
print(x)
```

# Boolean Indexing, Set Operations, and Sorting
### Boolean Indexing allows us to select elements using logical arguments instead of explicit indices
### Set operations are useful when comparing two numpy arrays

```python
x = np.arange(0,25).reshape(5,5)
x
```

```python
print(x[x>10], '\n')
print(x[x<=7], '\n')
print(x[(x>10) & (x<=17)], '\n')
```

```python
x[(x>10) & (x<=17)] = -1
```

```python
print(x)
```

```python
# Set Operations
x = np.array([1,2,3,4,5])
y = np.array([6,7,2,8,4])

print("The elements that are in both x and y:", np.intersect1d(x,y))
print("The elements that are in x but not in y:", np.setdiff1d(x,y))
print("All the elements of x and y combined:", np.union1d(x,y))
```

```python
# Sorting a numpy array using sort
# Used as a function, the array is sorted out of place
# Used as a method, the array is sorted in place

x = np.random.randint(1,11, (10,))
x
```

### Broadcasting

```python
print(w, '\n')
print('Add 3 to each element')
print( w + 3, '\n')
print('Subtract 3 from each element')
print( w - 3, '\n')
print('Multiply each element by 3')
print(w * 3, '\n')
print('Divide each element by 3')
print(w / 3, '\n')

# This codes above are executeable because of something called broadcasting
# Numpy is able to broadcast the smaller array along the bigger array. 
```

```python
x = np.arange(0,9).reshape(3,3)
x
```

```python
y = np.arange(0,3)
y
```

```python
# numpy broadcasts the smaller array along the bigger array. 
# It adds 0 to the first column of x, 1 to the 2nd column and 2 to the 3rd column
# This is only possible when the shape of the smaller array can be expanded to fit the shape of the bigger array

print(y+x) 
```
